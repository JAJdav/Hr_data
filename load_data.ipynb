```python
import pandas as pd
from sqlalchemy import create_engine
from sqlalchemy.exc import SQLAlchemyError
```

```python
df=pd.read_csv("hr_raw.csv")
```

```python
df.head()
```

```python
USER = "root"
PASSWORD = "123456"
HOST = "localhost"       
PORT = 3306              
DATABASE = "hr"


connection_string = f"mysql+pymysql://{USER}:{PASSWORD}@{HOST}:{PORT}/{DATABASE}"
engine = create_engine(connection_string)

try:

    df.to_sql(
        name="hr_data",     
        con=engine,
        if_exists="replace", 
        index=False         
    )
    print("✅ DataFrame successfully written to MySQL table 'hr'.")
except SQLAlchemyError as e:
    print("❌ Error writing to MySQL:", e)
finally:
    engine.dispose()  # Close connection
```
